<h1> Equiblibrium of Spruce Budworm

$ \frac{du}{dt} = Ru\left(1 - \frac{u}{Q} \right) - \frac{u^2}{1+u^2} = 0 $

```python
# import libraries
from scipy.optimize import newton
from matplotlib import pyplot as plt
import numpy as np
```

```python
# define my variable
R = 0.2
Q = 230
uo = 0.2
```

```python
# define our function
def f(u,R,Q):
  fx = R*u*(1-u/Q) - u**2/(1+u**2)
  return fx
```

```python
# Plot the function
u = np.arange(0.0001,19.99,0.001)
fu = []
for i in u:
  x = f(i,R,Q)
  fu.append(x)
```

```python
plt.plot(u,fu)
plt.grid()
plt.show()
```

```python
# Use Newton method to estimate the root
root = newton(f,uo,args=(R,Q))
print(root)
```
