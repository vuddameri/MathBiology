<h1> Ludwig–Jones–Holling spruce budworm model

$\frac{dB}{dt} = r_B \,B \left(1 - \frac{B}{K_b} \right) - \frac{\beta B^2}{\alpha^2 + B^2}$

B: Budworm density

t: time

$r_B$: maximum Budworm Growth Rate

$K_s$: Carrying capacity of the Spruce

$\beta$: Maximum Predation Rate of Budworm

$\alpha$: Budworm half saturation rate

<h1> Non-Dimensional Form


$ \frac{du}{d\tau} = R \times u \left(1-\frac{u}{Q}\right) - \frac{u^2}{1+u^2} $


$ u = \frac{B}{\alpha}$

$ \tau = \frac{\beta}{\alpha}t $

$ R = \frac{r_B\alpha}{\beta} $

$Q  = \frac{K_B}{\alpha}$

R is the relative budworm growth rate (relative to predation)

Q is the relative forest growth rate (relative to budworm saturation)

<h1> Code

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
import ipywidgets as widgets
from ipywidgets import interact
```

```python
# Dimensionless spruce budworm model
def spruce_budworm(tau, u, par):
  R = par[0]
  Q = par[1]
  dudtau = R*u*(1 - u/Q) - u**2/(1 + u**2)
  return dudtau
```

```python
# A function to solve ODE
def ode(R, Q):
  par = [R,Q]
  sol = solve_ivp(fun=spruce_budworm,y0=u0,t_span=tauspan,method='LSODA',t_eval = tau,args=(par,))
  return sol
```

```python
# Function to make the plot; This function returns none but creates a side effect of the plot
def plot(sol,R,Q):
  plt.figure(figsize=(8, 5))
  plt.plot(sol.t, sol.y[0], linewidth=2)
  plt.xlabel(r"Dimensionless Time, $\tau$")
  plt.ylabel(r"Dimensionless Budworm Population, $u$")
  plt.title(f"R = {R:.2f}, Q = {Q:.1f}")
  plt.grid()
  plt.show()
  return None
```

```python
# Define time and initial conditions
# Initial dimensionless budworm population
u0 = [0.2]
# Dimensionless time, tau
tau_start = 0
tau_end = 100
tauspan = [tau_start,tau_end]
tau = np.linspace(tau_start, tau_end, 1000)
```

```python
# Create a simulate function that solves ODE and makes a plot
def simulate(R=0.5, Q=10.0):
  sol = ode(R,Q)
  plot(sol,R,Q)
  return None
```

```python
interact(
    simulate,

    R=widgets.FloatSlider(
        value=0.5,
        min=0.05,
        max=1.0,
        step=0.05,
        description="R:"
    ),

    Q=widgets.FloatSlider(
        value=10.0,
        min=1.0,
        max=25.0,
        step=0.5,
        description="Q:"
    )
);
```
